# Python Loops & Conditional Statements

### Q1: 1 se 20 tak numbers print karein, lekin sirf EVEN (jift) numbers. (Hint: use % 2 == 0)

```python
for n in range(1, 21):
   # Agar number ko 2 se divide karne par remainder 0 bache (Even)
   if n %2== 0:
    print(n)
```

### Q2: 1 se 20 tak ODD numbers print karein using loop and if.

```python
for s in range(1, 21):
   # Agar number 2 se poora divide NAHI hota (Odd)
   if s % 2 !=0:
    print(s)
```

### Q3: 1 se 10 tak loop chalayen. Agar number 5 ho tou 'Five found!' print karein, warna number print karein.

```python
for f in range(1, 11):
  if f ==5:
    print("Five found")
  else:
    print(f)
```

### Q4: Ek list banayen [3, 8, 12, 1, 9]. Loop laga kar sirf wo numbers print karein jo 5 se baray (greater) hon.

### Q6: List [2, -5, 10, -8, 0] par loop lagayen. Agar number > 0 hai tou 'Positive', < 0 hai tou 'Negative' aur 0 hai tou 'Zero' print karein.

```python
num_list = [2, -5, 10, -8, 0]
for n in num_list:
  if n > 0:
    print(f"{n} is positive")
  elif n < 0 :
      print(f"{n} is negative")
  else:
        print(f"{n} is zero")
```

### Q7: 1 se 10 tak loop chalayen, lekin jab number 7 aaye tou loop ko `break` kar dein.

```python
for z in range(1, 11):
  if z == 7:
    print(z)
    break
```

### Q8: 1 se 10 tak loop chalayen, lekin number 4 ko `continue` (skip) kar dein.

```python
for c in range(1, 11):
  if c == 4:
    continue
  print(c)
```

### Q9: String 'python' par loop lagayen. Agar character vowel (a, e, i, o, u) ho tou usay print karein.

```python
vowels = ['a','e','i','o','u']
word = "python"
for c in word:
  if c in vowels:
    print(c)
```

### Q10: Ages ki list [12, 18, 25, 15] lein. Agar age 18 ya us se zyada ho tou 'Adult' print karein, warna 'Minor'.

```python
ages = [12,18,25,15]
for age in ages:
  if age >=18:
    print(f"Age {age}: Adult")
  else:
      print(f"Age {age}: Minor")
```

### Q14: List ['Ali', 'Bilal', 'Ahmed', 'Sana'] mein sirf wo naam print karein jo 'A' se shuru hotay hain.

### Q15: Ek variable count = 0 banayen. List ['apple', 'mango', 'apple', 'banana', 'apple'] par loop lagayen aur count karein ke 'apple' kitni baar aya hai.

### Q16: Marks ki list [45, 80, 30, 95, 40] par loop lagayen. Agar marks 50 se zyada hain tou 'Pass' print karein, warna 'Fail'.

### Q17: List [150, 45, 200, 99, 120] mein se sirf wo numbers print karein jo 100 aur 200 ke darmiyan hon.

```python
Numbers = [150, 45, 200, 99, 120]
for n in Numbers:
  if n >= 100 and n <= 200:
    print(f"The number b/w 100 and 200 is: {n}")
```

### Q18: 10 se 1 tak reverse loop chalayen. Jab number 5 aaye tou print karein 'Halfway there!'.

```python
for l in range(10, 0, -1):
  if l == 5:
    print("Halfway there")
  else:
    print(l)
```

### Q19: 1 se 10 tak tamam EVEN numbers ka sum (jama) calculate karein ek variable use kar ke, aur aakhir mein print karein.

```python
even_sum = 0
for s in range(1, 11):
  if s % 2 == 0:
    even_sum = even_sum + s

print("Tamam Even numbers ka total sum:", even_sum)
```

### Q20: Password check: List ['123', 'abc', 'admin123', 'xyz'] par loop lagayen. Agar 'admin123' mil jaye tou 'Access Granted' print kar ke loop break kar dein.

```python
passwords = ['123', 'abc', 'admin123', 'xyz']
for p in passwords:
  if p == 'admin123':
    print('Access Granted')
    break
```
